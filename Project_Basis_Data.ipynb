```python
# =====================================================
# IMPORT LIBRARY
# =====================================================

import sqlite3
import pandas as pd
```

```python
# =====================================================
# LOAD DATA CSV SMART PARKING
# =====================================================

from google.colab import drive
drive.mount('/content/gdrive')

# Membaca file CSV
data = pd.read_csv(
    "/content/gdrive/MyDrive/Basis Data/IIoT_Smart_Parking_Management.csv"
)

# Menampilkan 5 data pertama
print(data.head())
```

```python
# =====================================================
# KONEKSI SQL
# =====================================================

conn = sqlite3.connect("smart_parking_iiot.db")
cursor = conn.cursor()
```

```python
# =====================================================
# SIMPAN DATA KE SQL
# =====================================================

data.to_sql(
    "smart_parking",
    conn,
    if_exists="replace",
    index=False
)

print("Database Smart Parking berhasil dibuat!")
```

```python
# =====================================================
# FUNGSI MENJALANKAN QUERY
# =====================================================

def run_query_and_display(query_name, sql_query):

    print(f"\n{'=' * 70}")
    print(f"QUERY: {query_name}")
    print(f"{'=' * 70}")

    print(f"SQL:\n{sql_query}")

    try:
        result = pd.read_sql_query(sql_query, conn)

        print("\nHasil:")
        print(result.to_string(index=False))

    except Exception as e:
        print("ERROR:", e)
```

1. TOP OCCUPANCY RATE TERTINGGI

```python
run_query_and_display(
    "Top 10 Occupancy Rate Tertinggi",

    """
    SELECT
        Occupancy_Status,
        Occupancy_Rate,
        Parking_Duration
    FROM smart_parking
    ORDER BY Occupancy_Rate DESC
    LIMIT 10;
    """
)
```

Interpretasi:

Query ini digunakan untuk menampilkan 10 data dengan tingkat okupansi parkir tertinggi. Data yang ditampilkan meliputi status parkir, tingkat okupansi, dan durasi parkir. Query ini membantu mengetahui kondisi area parkir yang paling padat.

2. JUMLAH STATUS PARKIR

```python
run_query_and_display(
    "Jumlah Status Parkir",

    """
    SELECT
        Occupancy_Status,
        COUNT(*) AS jumlah_kendaraan
    FROM smart_parking
    GROUP BY Occupancy_Status
    ORDER BY jumlah_kendaraan DESC;
    """
)
```

Interpretasi:

Query ini digunakan untuk menghitung jumlah kendaraan berdasarkan status parkir, seperti occupied atau vacant. Hasil query dapat digunakan untuk mengetahui kondisi penggunaan area parkir secara keseluruhan.

3. JUMLAH KENDARAAN BERDASARKAN TIPE

```python
run_query_and_display(
    "Jumlah Kendaraan Berdasarkan Tipe",

    """
    SELECT
        Vehicle_Type,
        COUNT(*) AS total_kendaraan
    FROM smart_parking
    GROUP BY Vehicle_Type
    ORDER BY total_kendaraan DESC;
    """
)
```

Interpretasi:

Query ini digunakan untuk menghitung jumlah kendaraan berdasarkan jenis kendaraan, misalnya mobil, motor, atau kendaraan lainnya. Informasi ini membantu mengetahui jenis kendaraan yang paling banyak menggunakan area parkir

4. JUMLAH KENDARAAN LISTRIK

```python
run_query_and_display(
    "Jumlah Kendaraan Listrik",

    """
    SELECT
        Electric_Vehicle,
        COUNT(*) AS jumlah
    FROM smart_parking
    GROUP BY Electric_Vehicle;
    """
)
```

Interpretasi:

Query ini digunakan untuk mengetahui jumlah kendaraan listrik dan non-listrik pada sistem parkir. Data ini dapat digunakan untuk analisis penggunaan kendaraan ramah lingkungan.

5. RATA-RATA DURASI PARKIR

```python
run_query_and_display(
    "Rata-rata Durasi Parkir",

    """
    SELECT
        AVG(Parking_Duration) AS rata_rata_durasi
    FROM smart_parking;
    """
)
```

Interpretasi:

Query ini digunakan untuk menghitung rata-rata durasi kendaraan saat parkir. Hasil query membantu mengetahui berapa lama kendaraan biasanya menggunakan slot parkir.

6. TRAFFIC LEVEL PALING TINGGI

```python
run_query_and_display(
    "Traffic Level",

    """
    SELECT
        Nearby_Traffic_Level,
        COUNT(*) AS total
    FROM smart_parking
    GROUP BY Nearby_Traffic_Level
    ORDER BY total DESC;
    """
)
```

Interpretasi:

Query ini digunakan untuk melihat tingkat lalu lintas di sekitar area parkir berdasarkan kategori traffic level. Hasil query membantu mengetahui kondisi lalu lintas yang paling sering terjadi.

Query ini digunakan untuk menghitung total keseluruhan pembayaran parkir yang dilakukan pengguna kendaraan.

Interpretasi:

12. PENGARUH CUACA TERHADAP OCCUPANCY

```python
run_query_and_display(
    "Pengaruh Cuaca terhadap Occupancy",

    """
    SELECT
        Weather_Temperature,
        AVG(Occupancy_Rate) AS rata_occupancy
    FROM smart_parking
    GROUP BY Weather_Temperature
    ORDER BY rata_occupancy DESC;
    """
)
```

Interpretasi:

Query ini digunakan untuk melihat hubungan antara suhu cuaca dengan rata-rata tingkat okupansi parkir. Hasil query membantu mengetahui apakah kondisi cuaca memengaruhi penggunaan area parkir
